## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

**Research question:**

How can multiple content performance and search-related signals be combined to identify which pages should be prioritized for human review and potential improvement?

**Decision:** The project aims to help SEO and content teams decide which pages require attention first when time and editorial resources are limited.

**Who takes action:** SEO specialists, content strategists, and editorial teams.

**Possible actions:**
- Review pages with declining performance.
- Update outdated or underperforming content.
- Improve page titles, metadata, and content structure.
- Investigate pages with unusual performance patterns.
- Continue monitoring pages that do not require immediate intervention.

**Cost of a wrong recommendation:**

A false positive may lead to unnecessary content updates, wasted working hours, and inefficient use of editorial resources. A false negative may cause a valuable content opportunity or an important performance decline to be overlooked.

The proposed system will provide decision support rather than make automatic content changes. Final decisions will remain with human specialists.

### Initial data observations

The dataset contains **30,000 records representing 30,000 unique content pages**.

The initial exploration reveals several notable patterns:

- **54.21% (16,262 pages)** are classified as `down`, indicating that more than half of the pages belong to the downward-trend category.
- **19.87% (5,962 pages)** are classified as `stable`, while only **14.63% (4,388 pages)** are classified as `up`.
- Median impressions vary considerably across trend categories: `stable` pages have a median of **1,944.5 impressions**, compared with **961** for `down` pages, **587** for `up` pages, **4** for `flat` pages, and **3** for `new` pages.

These observations suggest that content pages have different performance patterns and that a single metric may not be sufficient to prioritize them effectively. The relatively large proportion of pages classified as `down` makes it worthwhile to investigate which pages may require further human review.

These are initial descriptive findings, not evidence that a particular content characteristic causes performance changes. The meaning of each trend category should also be interpreted within the dataset's definitions.
